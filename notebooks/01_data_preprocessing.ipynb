```python
results = results_raw.iloc[1:].copy()

results.columns = [
    "slno", "L", "m", "perf", "dia",
    "blank1", "blank2", "blank3", "blank4",
    "E", "Fmax", "Tm", "def",
    "destntype", "place", "spot"
]

results = results.drop(columns=["blank1", "blank2", "blank3", "blank4"])

results.head()
```

```python
# Clean diameter values
results["dia"] = (
    results["dia"]
    .astype(str)
    .str.replace(r"\\n", "", regex=True)
    .str.strip()
)

# Convert diameter to numeric
results["dia"] = pd.to_numeric(results["dia"], errors="coerce")

results.head()
```
